### D13. Read a file line by line and split each line

`records_sample.txt` (in this `LAB` folder) has one record per line, like `R-001,87,100`.

1. Open it with `with` and loop over it line by line.

### D16. A list of dicts

1. Create a list of three dicts:
   ```python
   records = [
       {"label": "R-001", "value": 87.0, "limit": 100.0},
       {"label": "R-002", "value": 120.0, "limit": 100.0},
       {"label": "R-003", "value": 45.0, "limit": 100.0},
   ]
   ```
2. Loop over `records`. For each one, print its label and `OVER LIMIT` if `value` is greater than
   `limit`, otherwise `OK`.

**Expected output:**
```
R-001: OK
R-002: OVER LIMIT
R-003: OK
```

````python
# D16

# Create a list of dictionaries containing record details
records = [
       {"label": "R-001", "value": 87.0, "limit": 100.0},
       {"label": "R-002", "value": 120.0, "limit": 100.0},
       {"label": "R-003", "value": 45.0, "limit": 100.0},
       ]

# Loop through each record in the list   
for record in records:
    
    # Extract the label, value and limit from the dictionary
    label = record['label']
    value = record['value']
    limit = record['limit']
    
    if value > limit:                       # Check whether the value exceeds the limit
        print(f"{label}: OVER LIMIT")
    else:
        print(f"{label}: OK")
````

### C1. Load records safely from a file

Write `load_records(path)` with a one-line docstring. It reads `label,value,limit` lines from a file
and returns a **list of dicts**. If the file does not exist, it prints a clear message and returns an
empty list instead of crashing.

Then load `records_sample.txt` and print each record's label and status, as in D16.

**Test it:**

| Call | Expected output |
|---|---|
| `load_records("records_sample.txt")` | the same three lines as D16 |
| `load_records("nope.txt")` | a clear message, then `[]` |

````python
# C1

def load_records(path):
    """Reads label,value,limit from a file and returns a list of dicts""" 
    try:
        with open(path) as f:
            new = []                #empty list to add the dictonary values
            for line in f:
                label, value, limit = line.strip().split(',')   #seperating the line into label, value, limit
                new.append({'label': label,                     #appending the values to the dictionary
                            'value': value,
                            'limit': limit,
                            })
            return new
    except FileNotFoundError:
        print('The file was not found')
        return []     
    
records = load_records('records_sample.txt')     #storing the records_sample in a variable
for r in records:                                
    label = r['label']
    value = int(r['value'])         #converting the value to data type int
    limit = int(r['limit'])         #converting the value to data type int
    
    if value > limit:               #checking if value is greater than limit
        print(f"{label}: OVER LIMIT")
    else:
        print(f"{label}: OK")

load_records('none.txt')            #calling the function with none.txt file
````

### C2. Write the results to a file

Using your records from C1, write one line per record to a new file called `results.txt`,
in the form `label,status`. Then read `results.txt` back and print it.

R-003,OK

````python
# C2
with open('results.txt', 'w')as f:
     """Write each record's label and status to a file."""
     for r in records:
        label = r['label']
        value = int(r['value'])         #converting value to data type int
        limit = int(r['limit'])         #converting value to data type int
        
        if value > limit:               #checking if value is greater than limit
            status = 'OVER LIMIT'
        else:
            status = 'OK'
        f.write(f"{label},{status}\n")
        
with open('results.txt')as x:           # Open the file in read mode and display its contents
    print(x.read())
                
                
````
